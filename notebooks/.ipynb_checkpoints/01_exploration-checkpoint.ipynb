# 6. Initial Data Profiling
Profiling data types, missing values, and uniqueness.

```python
def profile_dataframe(df):
    rows = []

    for column in df.columns:
        series = df[column]

        try:
            unique_count = series.nunique(dropna=True)
        except TypeError:
            unique_count = series.astype(str).nunique(dropna=True)

        rows.append({
            "column": column,
            "dtype": str(series.dtype),
            "non_null": series.notna().sum(),
            "null_count": series.isna().sum(),
            "null_pct": round(series.isna().mean() * 100, 2),
            "unique_count": unique_count
        })

    return (
        pd.DataFrame(rows)
        .sort_values("null_pct", ascending=False)
        .reset_index(drop=True)
    )
```

```python
checks = {}

checks["approval_before_purchase"] = (
    orders["order_approved_at"] <
    orders["order_purchase_timestamp"]
)

checks["carrier_before_purchase"] = (
    orders["order_delivered_carrier_date"] <
    orders["order_purchase_timestamp"]
)

checks["delivery_before_purchase"] = (
    orders["order_delivered_customer_date"] <
    orders["order_purchase_timestamp"]
)

checks["delivery_before_carrier"] = (
    orders["order_delivered_customer_date"] <
    orders["order_delivered_carrier_date"]
)

validation_summary = pd.DataFrame([
    {
        "rule": name,
        "violations": condition.sum(),
        "violation_pct": round(condition.mean() * 100, 4)
    }
    for name, condition in checks.items()
])

display(validation_summary)
```

# 5. Source / Entity Map
Analyzing referential integrity and how tables map to one another.

```python
relationships = {
    "orders → customers": (
        orders["customer_id"].isin(customers["customer_id"])
    ),
    
    "order_items → orders": (
        order_items["order_id"].isin(orders["order_id"])
    ),
    
    "order_items → products": (
        order_items["product_id"].isin(products["product_id"])
    ),
    
    "order_items → sellers": (
        order_items["seller_id"].isin(sellers["seller_id"])
    ),
    
    "payments → orders": (
        payments["order_id"].isin(orders["order_id"])
    ),
    
    "reviews → orders": (
        reviews["order_id"].isin(orders["order_id"])
    )
}

relationship_results = pd.DataFrame([
    {
        "relationship": name,
        "rows_checked": len(condition),
        "matched": condition.sum(),
        "unmatched": (~condition).sum(),
        "unmatched_pct": round((~condition).mean() * 100, 4)
    }
    for name, condition in relationships.items()
])

display(relationship_results)
```

# 10. Feature Engineering / Order-Level Modeling
Aggregating child tables (items, payments, reviews) up to the order grain to prevent many-to-many explosions.

```python
order_item_counts = (
    order_items
    .groupby("order_id")
    .agg(
        item_rows=("order_item_id", "count"),
        distinct_products=("product_id", "nunique"),
        distinct_sellers=("seller_id", "nunique")
    )
    .reset_index()
)

display(order_item_counts.describe())
```

```python
seller_distribution = (
    order_item_counts["distinct_sellers"]
    .value_counts()
    .sort_index()
)

display(seller_distribution)
```

```python
multi_seller_orders = order_item_counts[
    order_item_counts["distinct_sellers"] > 1
]

print("Orders with multiple sellers:", len(multi_seller_orders))
print(
    "Percentage:",
    round(
        len(multi_seller_orders) /
        len(order_item_counts) * 100,
        2
    ),
    "%"
)

display(multi_seller_orders.head(10))
```

```python
payment_counts = (
    payments
    .groupby("order_id")
    .agg(
        payment_rows=("payment_sequential", "count"),
        payment_methods=("payment_type", "nunique"),
        total_payment=("payment_value", "sum")
    )
    .reset_index()
)

display(payment_counts.describe())
```

```python
multiple_payment_orders = payment_counts[
    payment_counts["payment_rows"] > 1
]

print(
    "Orders with multiple payment records:",
    len(multiple_payment_orders)
)

print(
    "Percentage:",
    round(
        len(multiple_payment_orders) /
        len(payment_counts) * 100,
        2
    ),
    "%"
)

display(multiple_payment_orders.head(10))
```

```python
review_counts = (
    reviews
    .groupby("order_id")
    .agg(
        review_rows=("review_id", "count"),
        distinct_review_ids=("review_id", "nunique"),
        average_score=("review_score", "mean")
    )
    .reset_index()
)

display(review_counts.describe())
```

```python
multiple_review_orders = review_counts[
    review_counts["review_rows"] > 1
]

print(
    "Orders with multiple review records:",
    len(multiple_review_orders)
)

print(
    "Percentage:",
    round(
        len(multiple_review_orders) /
        len(review_counts) * 100,
        2
    ),
    "%"
)
```

```python
order_features = orders.copy()

order_features = order_features.merge(
    order_item_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

order_features = order_features.merge(
    payment_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

order_features = order_features.merge(
    review_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(order_features))
print("Columns:", len(order_features.columns))

display(order_features.head())
```

```python
order_features["purchase_to_carrier_days"] = (
    order_features["order_delivered_carrier_date"]
    - order_features["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 3600)

order_features["carrier_to_delivery_days"] = (
    order_features["order_delivered_customer_date"]
    - order_features["order_delivered_carrier_date"]
).dt.total_seconds() / (24 * 3600)

order_features["delivery_variance_days"] = (
    order_features["order_delivered_customer_date"]
    - order_features["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 3600)


order_features["is_late"] = (
    order_features["delivery_variance_days"] > 0
)
```

```python
order_features["is_late"] = pd.Series(
    pd.NA,
    index=order_features.index,
    dtype="boolean"
)

valid_delivery = (
    order_features["order_status"].eq("delivered") &
    order_features["order_delivered_customer_date"].notna() &
    order_features["order_estimated_delivery_date"].notna() &
    order_features["delivery_variance_days"].notna()
)

order_features.loc[valid_delivery, "is_late"] = (
    order_features.loc[valid_delivery, "delivery_variance_days"] > 0
)
```

# 13. Operational Segmentation
Analyzing performance segmented by seller delivery volume.

```python
seller_order_features = (
    order_features.loc[
        validated &
        order_features["distinct_sellers"].notna()
    ][
        [
            "order_id",
            "is_late",
            "purchase_to_carrier_days",
            "carrier_to_delivery_days"
        ]
    ]
    .merge(
        order_items[
            ["order_id", "seller_id"]
        ].drop_duplicates(),
        on="order_id",
        how="left"
    )
)

seller_performance = (
    seller_order_features
    .groupby("seller_id")
    .agg(
        delivered_orders=("order_id", "nunique"),
        late_orders=("is_late", "sum"),
        late_delivery_rate=("is_late", "mean"),
        median_purchase_to_carrier=(
            "purchase_to_carrier_days",
            "median"
        ),
        median_carrier_to_delivery=(
            "carrier_to_delivery_days",
            "median"
        )
    )
    .reset_index()
)

display(seller_performance.head())
```

```python
display(
    seller_performance["delivered_orders"]
    .describe()
)
```

```python
thresholds = [5, 10, 25, 50, 100]

seller_volume_sensitivity = []

total_sellers = len(seller_performance)
total_delivered = seller_performance["delivered_orders"].sum()

for threshold in thresholds:
    eligible = seller_performance[
        seller_performance["delivered_orders"] >= threshold
    ]

    seller_volume_sensitivity.append({
        "minimum_delivered_orders": threshold,
        "sellers_remaining": len(eligible),
        "seller_coverage_pct": (
            len(eligible) / total_sellers * 100
        ),
        "delivered_orders_covered": (
            eligible["delivered_orders"].sum()
        ),
        "delivery_order_coverage_pct": (
            eligible["delivered_orders"].sum()
            / total_delivered * 100
        )
    })

seller_volume_sensitivity = pd.DataFrame(
    seller_volume_sensitivity
)

display(seller_volume_sensitivity)
```

```python
seller_analysis = seller_performance[
    seller_performance["delivered_orders"] >= 10
].copy()

print("Sellers:", len(seller_analysis))
print(
    "Delivered orders:",
    seller_analysis["delivered_orders"].sum()
)
print(
    "Order coverage:",
    round(
        seller_analysis["delivered_orders"].sum()
        / seller_performance["delivered_orders"].sum()
        * 100,
        2
    ),
    "%"
)
```

# 14. Key Findings
- Delivery delays are associated with substantially longer carrier-to-customer durations.
- Missing timestamps and lifecycle anomalies (e.g. carrier before purchase) are present in the raw data and must be excluded from timing metrics.
- Aggregation is required for items, payments, and reviews before joining to the order grain to prevent row multiplication.

# 15. Assumptions and Limitations
- The minimum seller threshold of 10 validated orders is an analytical assumption, not a data quality rule. It should remain configurable.
- Large positive durations are heavy-tailed but logically possible; they are treated as valid unless they violate chronology.
- We are evaluating association, not causal attribution (e.g., we do not claim the carrier *caused* the delay).

# 16. Pipeline Design Handoff
- Extract: Preserve raw source tables.
- Validate: Assert required schema, uniqueness, referential integrity, and logical chronology.
- Clean: Explicitly document rules (e.g. flagging anomalies without silently dropping records).
- Transform: Construct order-level features (1 row per order).
- Publish: Output final KPIs safely.
