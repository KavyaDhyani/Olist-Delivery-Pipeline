# Olist E-Commerce Operations — Data Exploration

## Objective

Explore the Olist e-commerce dataset to determine whether the available
source tables can support a dependable operational workflow for analysing
order fulfilment and delivery performance.

This notebook is exploratory.

The goals are to:

1. Understand the available source tables
2. Identify the grain of each table
3. Identify candidate primary and foreign keys
4. Understand relationships between entities
5. Profile data quality
6. Identify meaningful business validation rules
7. Reconstruct the order lifecycle
8. Identify candidate operational KPIs
9. Document assumptions, unknowns and limitations

No raw source data will be silently modified in this notebook.

## Dataset Grain

Before joining anything, determine what one row represents in each source.

Examples of possible grains:

- one customer
- one order
- one order item
- one payment associated with an order
- one review associated with an order
- one product
- one seller
- one geolocation record

```python
for name, df in datasets.items():
    print("=" * 80)
    print(name)
    print("Shape:", df.shape)
    print("Columns:")
    
    for column in df.columns:
        print(f"  - {column}")
```

```python
def profile_dataframe(df):
    rows = []

    for column in df.columns:
        series = df[column]

        try:
            unique_count = series.nunique(dropna=True)
        except TypeError:
            unique_count = series.astype(str).nunique(dropna=True)

        rows.append({
            "column": column,
            "dtype": str(series.dtype),
            "non_null": series.notna().sum(),
            "null_count": series.isna().sum(),
            "null_pct": round(series.isna().mean() * 100, 2),
            "unique_count": unique_count
        })

    return (
        pd.DataFrame(rows)
        .sort_values("null_pct", ascending=False)
        .reset_index(drop=True)
    )
```

```python
checks = {}

checks["approval_before_purchase"] = (
    orders["order_approved_at"] <
    orders["order_purchase_timestamp"]
)

checks["carrier_before_purchase"] = (
    orders["order_delivered_carrier_date"] <
    orders["order_purchase_timestamp"]
)

checks["delivery_before_purchase"] = (
    orders["order_delivered_customer_date"] <
    orders["order_purchase_timestamp"]
)

checks["delivery_before_carrier"] = (
    orders["order_delivered_customer_date"] <
    orders["order_delivered_carrier_date"]
)

validation_summary = pd.DataFrame([
    {
        "rule": name,
        "violations": condition.sum(),
        "violation_pct": round(condition.mean() * 100, 4)
    }
    for name, condition in checks.items()
])

display(validation_summary)
```

# Candidate KPI

## Primary KPI hypothesis

Delivery reliability.

Potential primary measure:

    percentage of delivered orders that were delivered after
    the estimated delivery date.

Supporting metrics may include:

- median delivery delay
- average delivery delay
- order approval time
- carrier handoff time
- seller-level late-delivery rate
- review score associated with late deliveries

These are candidate metrics only and will be finalized after
completing validation and workflow analysis.

```python
order_item_counts = (
    order_items
    .groupby("order_id")
    .agg(
        item_rows=("order_item_id", "count"),
        distinct_products=("product_id", "nunique"),
        distinct_sellers=("seller_id", "nunique")
    )
    .reset_index()
)

display(order_item_counts.describe())
```

```python
payment_features = (
    payments
    .groupby("order_id")
    .agg(
        payment_count=("payment_sequential", "count"),
        payment_method_count=("payment_type", "nunique"),
        total_payment=("payment_value", "sum"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)

display(payment_features.head())
print("Rows:", len(payment_features))
```

```python
review_features = (
    reviews
    .groupby("order_id")
    .agg(
        review_count=("review_id", "count"),
        average_review_score=("review_score", "mean")
    )
    .reset_index()
)

display(review_features.head())
print("Rows:", len(review_features))
```

```python
order_features = orders.copy()

order_features = order_features.merge(
    order_item_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

order_features = order_features.merge(
    payment_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

order_features = order_features.merge(
    review_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(order_features))
print("Columns:", len(order_features.columns))

display(order_features.head())
```

```python
print("Expected rows:", len(orders))
print("Actual rows:", len(order_features))
print("Unique order IDs:", order_features["order_id"].nunique())
print(
    "Duplicate order IDs:",
    order_features["order_id"].duplicated().sum()
)
```

```python
feature_coverage = pd.DataFrame({
    "feature_group": [
        "Order items",
        "Payments",
        "Reviews"
    ],
    "orders_with_data": [
        order_features["item_count"].notna().sum(),
        order_features["payment_count"].notna().sum(),
        order_features["review_count"].notna().sum()
    ],
    "orders_without_data": [
        order_features["item_count"].isna().sum(),
        order_features["payment_count"].isna().sum(),
        order_features["review_count"].isna().sum()
    ]
})

feature_coverage["coverage_pct"] = (
    feature_coverage["orders_with_data"]
    / len(order_features)
    * 100
)

display(feature_coverage)
```

```python
order_features["purchase_to_carrier_days"] = (
    order_features["order_delivered_carrier_date"]
    - order_features["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 3600)

order_features["carrier_to_delivery_days"] = (
    order_features["order_delivered_customer_date"]
    - order_features["order_delivered_carrier_date"]
).dt.total_seconds() / (24 * 3600)

order_features["delivery_variance_days"] = (
    order_features["order_delivered_customer_date"]
    - order_features["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 3600)


order_features["is_late"] = (
    order_features["delivery_variance_days"] > 0
)
```

```python
order_features["is_late"] = pd.Series(
    pd.NA,
    index=order_features.index,
    dtype="boolean"
)

valid_delivery = (
    order_features["order_status"].eq("delivered") &
    order_features["order_delivered_customer_date"].notna() &
    order_features["order_estimated_delivery_date"].notna() &
    order_features["delivery_variance_days"].notna()
)

order_features.loc[valid_delivery, "is_late"] = (
    order_features.loc[valid_delivery, "delivery_variance_days"] > 0
)
```

```python
display(
    order_features[
        [
            "order_id",
            "order_status",
            "purchase_to_carrier_days",
            "carrier_to_delivery_days",
            "delivery_variance_days",
            "is_late"
        ]
    ].head(10)
)
```

```python
print("Late:", (order_features["is_late"] == True).sum())
print("Not late:", (order_features["is_late"] == False).sum())
print("Unknown:", order_features["is_late"].isna().sum())
```

```python
customer_features = customers[
    [
        "customer_id",
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ]
].copy()

order_features = order_features.merge(
    customer_features,
    on="customer_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(order_features))
print("Unique orders:", order_features["order_id"].nunique())
```

```python
customer_coverage = pd.DataFrame({
    "field": [
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ],
    "missing": [
        order_features["customer_unique_id"].isna().sum(),
        order_features["customer_city"].isna().sum(),
        order_features["customer_state"].isna().sum()
    ]
})

display(customer_coverage)
```

```python
item_categories = order_items[
    ["order_id", "product_id"]
].merge(
    products[
        ["product_id", "product_category_name"]
    ],
    on="product_id",
    how="left",
    validate="many_to_one"
)

print("Item rows:", len(item_categories))
print(
    "Missing categories:",
    item_categories["product_category_name"].isna().sum()
)

display(item_categories.head())
```

```python
category_translation = pd.read_csv(
    RAW_DIR / "product_category_name_translation.csv"
)

print("Rows:", len(category_translation))
print("Columns:", category_translation.columns.tolist())
display(category_translation.head())
```

```python
item_categories = item_categories.merge(
    category_translation,
    on="product_category_name",
    how="left",
    validate="many_to_one"
)

display(
    item_categories[
        [
            "order_id",
            "product_id",
            "product_category_name",
            "product_category_name_english"
        ]
    ].head(10)
)
```

```python
category_features = (
    item_categories
    .groupby("order_id")
    .agg(
        distinct_categories=(
            "product_category_name_english",
            "nunique"
        )
    )
    .reset_index()
)

display(category_features.head())
```

```python
order_features = order_features.merge(
    category_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(order_features))
print("Unique orders:", order_features["order_id"].nunique())
print("Duplicate order IDs:", order_features["order_id"].duplicated().sum())
```

```python
item_seller_geo = order_items[
    ["order_id", "seller_id"]
].merge(
    sellers[
        ["seller_id", "seller_city", "seller_state"]
    ],
    on="seller_id",
    how="left",
    validate="many_to_one"
)

display(item_seller_geo.head())
```

```python
seller_geo_features = (
    item_seller_geo
    .groupby("order_id")
    .agg(
        distinct_seller_states=("seller_state", "nunique"),
        distinct_seller_cities=("seller_city", "nunique")
    )
    .reset_index()
)

display(seller_geo_features.head())
```

```python
order_features = order_features.merge(
    seller_geo_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(order_features))
print("Unique orders:", order_features["order_id"].nunique())
print("Duplicate order IDs:", order_features["order_id"].duplicated().sum())
```

```python
delivery_summary = {
    "total_orders": len(order_features),
    "delivered_orders": (
        order_features["order_status"] == "delivered"
    ).sum(),
    "validated_delivered_orders": (
        order_features["is_late"].notna()
    ).sum(),
    "late_orders": (
        order_features["is_late"] == True
    ).sum(),
    "on_time_orders": (
        order_features["is_late"] == False
    ).sum(),
}

delivery_summary
```

```python
validated = order_features["is_late"].notna()

baseline_kpis = pd.DataFrame({
    "metric": [
        "Late delivery rate",
        "Median delivery variance (days)",
        "Median purchase → carrier (days)",
        "Median carrier → customer (days)"
    ],
    "value": [
        order_features.loc[validated, "is_late"].mean() * 100,
        order_features.loc[validated, "delivery_variance_days"].median(),
        order_features.loc[validated, "purchase_to_carrier_days"].median(),
        order_features.loc[validated, "carrier_to_delivery_days"].median()
    ]
})

display(baseline_kpis)
```

```python
stage_summary = pd.DataFrame({
    "stage": [
        "Purchase → Carrier",
        "Carrier → Customer"
    ],
    "median_days": [
        order_features.loc[
            validated,
            "purchase_to_carrier_days"
        ].median(),
        order_features.loc[
            validated,
            "carrier_to_delivery_days"
        ].median()
    ],
    "mean_days": [
        order_features.loc[
            validated,
            "purchase_to_carrier_days"
        ].mean(),
        order_features.loc[
            validated,
            "carrier_to_delivery_days"
        ].mean()
    ],
    "orders": [
        order_features.loc[
            validated,
            "purchase_to_carrier_days"
        ].notna().sum(),
        order_features.loc[
            validated,
            "carrier_to_delivery_days"
        ].notna().sum()
    ]
})

display(stage_summary)
```

```python
stage_by_lateness = (
    order_features.loc[validated]
    .groupby("is_late")
    .agg(
        orders=("order_id", "count"),
        median_purchase_to_carrier=(
            "purchase_to_carrier_days",
            "median"
        ),
        median_carrier_to_delivery=(
            "carrier_to_delivery_days",
            "median"
        ),
        mean_purchase_to_carrier=(
            "purchase_to_carrier_days",
            "mean"
        ),
        mean_carrier_to_delivery=(
            "carrier_to_delivery_days",
            "mean"
        )
    )
    .reset_index()
)

display(stage_by_lateness)
```

```python
seller_order_features = (
    order_features.loc[
        validated &
        order_features["distinct_sellers"].notna()
    ][
        [
            "order_id",
            "is_late",
            "purchase_to_carrier_days",
            "carrier_to_delivery_days"
        ]
    ]
    .merge(
        order_items[
            ["order_id", "seller_id"]
        ].drop_duplicates(),
        on="order_id",
        how="left"
    )
)

seller_performance = (
    seller_order_features
    .groupby("seller_id")
    .agg(
        delivered_orders=("order_id", "nunique"),
        late_orders=("is_late", "sum"),
        late_delivery_rate=("is_late", "mean"),
        median_purchase_to_carrier=(
            "purchase_to_carrier_days",
            "median"
        ),
        median_carrier_to_delivery=(
            "carrier_to_delivery_days",
            "median"
        )
    )
    .reset_index()
)

display(seller_performance.head())
```

```python
display(
    seller_performance["delivered_orders"]
    .describe()
)
```

```python
thresholds = [5, 10, 25, 50, 100]

seller_volume_sensitivity = []

total_sellers = len(seller_performance)
total_delivered = seller_performance["delivered_orders"].sum()

for threshold in thresholds:
    eligible = seller_performance[
        seller_performance["delivered_orders"] >= threshold
    ]

    seller_volume_sensitivity.append({
        "minimum_delivered_orders": threshold,
        "sellers_remaining": len(eligible),
        "seller_coverage_pct": (
            len(eligible) / total_sellers * 100
        ),
        "delivered_orders_covered": (
            eligible["delivered_orders"].sum()
        ),
        "delivery_order_coverage_pct": (
            eligible["delivered_orders"].sum()
            / total_delivered * 100
        )
    })

seller_volume_sensitivity = pd.DataFrame(
    seller_volume_sensitivity
)

display(seller_volume_sensitivity)
```

```python
seller_analysis = seller_performance[
    seller_performance["delivered_orders"] >= 10
].copy()

print("Sellers:", len(seller_analysis))
print(
    "Delivered orders:",
    seller_analysis["delivered_orders"].sum()
)
print(
    "Order coverage:",
    round(
        seller_analysis["delivered_orders"].sum()
        / seller_performance["delivered_orders"].sum()
        * 100,
        2
    ),
    "%"
)
```
